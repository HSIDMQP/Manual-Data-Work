```python
import pandas as pd

data = pd.read_csv("endava_data_full.csv")
raw = pd.read_csv("RAW_endava_data_full.csv")
print(data["Document"].head())
print(raw["body"].head())
```

```python
row_count = len(data)
print(f"The data has {row_count} rows pre-cleaning")
```

```python
data = data.loc[
    data["Document"].fillna("").str.split().str.len() >= 50
].copy()

print(data.head())
```

```python
row_count = len(data)
print(f"The dataset has {row_count} rows with 50 or more words")
```

```python
# count unique values in each column
def count_unique_values(df, excluded_columns = None):
    excluded_columns = excluded_columns or []

    for column in df.columns:
        if column not in excluded_columns:
            print(f"\n{df[column].value_counts(dropna = False)}")

count_unique_values(
    data,
    excluded_columns = ["Document"]
)
```

```python
data.to_csv("endava_data_relevant.csv", index = False)
```

```python
sampled_data = data.groupby("Topic_group", group_keys = False).sample(
    n = 10,
    random_state = 75
)

print(sampled_data.head())
sampled_data.to_csv("endava_data_n10_samples.csv", index = False)
```

```python
sampled_data = data.groupby("Topic_group", group_keys = False).sample(
    n = 5,
    random_state = 75
)

print(sampled_data.head())
sampled_data.to_csv("endava_data_n5_samples.csv", index = False)
```
