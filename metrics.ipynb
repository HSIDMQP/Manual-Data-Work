```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support
)

run1 = pd.read_csv("datasets/copilot results/copilot_n5_p3_results_run1.csv")
run2 = pd.read_csv("datasets/copilot results/copilot_n5_p3_results_run2.csv")
run3 = pd.read_csv("datasets/copilot results/copilot_n5_p3_results_run3.csv")
run4 = pd.read_csv("datasets/copilot results/copilot_n5_p3_results_run4.csv")
print(run1.head())
```

```python
runs = {
    "run1": run1,
    "run2": run2,
    "run3": run3,
    "run4": run4
}

results = []

for name, df in runs.items():
    labels = df[["Topic_group", "Predicted_topic"]].dropna().astype("string")
    labels = labels.apply(lambda col: col.str.strip())
    labels = labels[
        labels["Topic_group"].ne("") & labels["Predicted_topic"].ne("")
    ]

    y_true = labels["Topic_group"]
    y_pred = labels["Predicted_topic"]

    row = {"file": name, "rows_scored": len(labels)}

    if len(labels):
        row["accuracy"] = accuracy_score(y_true, y_pred)

        for average in ("weighted", "macro"):
            precision, recall, f1, _ = precision_recall_fscore_support(
                y_true,
                y_pred,
                average = average,
                zero_division = 0
            )
            row[f"{average}_precision"] = precision
            row[f"{average}_recal"] = recall
            row[f"{average}_f1"] = f1


    results.append(row)

metrics_by_file = pd.DataFrame(results)
print(metrics_by_file)
```
